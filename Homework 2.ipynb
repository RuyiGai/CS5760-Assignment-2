# Part I--Q5

```python
import numpy as np

# Rows = System predictions
# Columns = Gold labels
cm = np.array([
    [5, 10, 5],    # Cat
    [15, 20, 10],  # Dog
    [0, 15, 10]    # Rabbit
])

classes = ["Cat", "Dog", "Rabbit"]

precisions = []
recalls = []

# Per-class precision and recall
for i, class_name in enumerate(classes):
    TP = cm[i, i]
    FP = np.sum(cm[i, :]) - TP
    FN = np.sum(cm[:, i]) - TP

    precision = TP / (TP + FP)
    recall = TP / (TP + FN)

    precisions.append(precision)
    recalls.append(recall)

    print(f"{class_name}:")
    print(f"  Precision = {precision:.4f}")
    print(f"  Recall    = {recall:.4f}")

# Macro averaging
macro_precision = np.mean(precisions)
macro_recall = np.mean(recalls)

print("\nMacro-Averaged Metrics:")
print(f"  Precision = {macro_precision:.4f}")
print(f"  Recall    = {macro_recall:.4f}")

# Micro averaging
total_TP = np.trace(cm)
total_FP = np.sum(cm) - total_TP
total_FN = np.sum(cm) - total_TP

micro_precision = total_TP / (total_TP + total_FP)
micro_recall = total_TP / (total_TP + total_FN)

print("\nMicro-Averaged Metrics:")
print(f"  Precision = {micro_precision:.4f}")
print(f"  Recall    = {micro_recall:.4f}")
```

# Part II--Q1

```python
from collections import Counter

# 1. Training corpus
corpus = [
    "<s> I love NLP </s>",
    "<s> I love deep learning </s>",
    "<s> deep learning is fun </s>"
]

# Convert sentences into lists of tokens
tokenized_corpus = [sentence.split() for sentence in corpus]

# 2. Compute unigram and bigram counts
unigram_counts = Counter()
bigram_counts = Counter()

for sentence in tokenized_corpus:
    # Unigram counts
    unigram_counts.update(sentence)

    # Bigram counts
    for i in range(len(sentence) - 1):
        bigram = (sentence[i], sentence[i + 1])
        bigram_counts[bigram] += 1

# Print unigram counts
print("Unigram Counts:")
for word, count in unigram_counts.items():
    print(f"{word}: {count}")

# Print bigram counts
print("\nBigram Counts:")
for bigram, count in bigram_counts.items():
    print(f"{bigram}: {count}")

# 3. Estimate bigram probabilities using MLE
# MLE formula: P(w2 | w1) = Count(w1, w2) / Count(w1)
bigram_probs = {}

for (w1, w2), count in bigram_counts.items():
    bigram_probs[(w1, w2)] = count / unigram_counts[w1]

print("\nBigram Probabilities:")
for bigram, probability in bigram_probs.items():
    print(f"P({bigram[1]} | {bigram[0]}) = {probability:.4f}")


# 4. Function to calculate sentence probability
def sentence_probability(sentence):
    tokens = sentence.split()
    # Initialize the cumulative probability to 1
    probability = 1.0

    for i in range(len(tokens) - 1):
        bigram = (tokens[i], tokens[i + 1])

        # If the bigram was never seen in the training corpus,
        # its MLE probability is 0.
        if bigram not in bigram_probs:
            return 0.0

        probability *= bigram_probs[bigram]

    return probability


# 5. Test the two sentences
sentence1 = "<s> I love NLP </s>"
sentence2 = "<s> I love deep learning </s>"

prob1 = sentence_probability(sentence1)
prob2 = sentence_probability(sentence2)

print("\nSentence Probabilities:")
print(f"{sentence1}: {prob1:.4f}")
print(f"{sentence2}: {prob2:.4f}")


# 6. Print which sentence the model prefers
if prob1 > prob2:
    print("\nThe model prefers Sentence 1.")
    print("Reason: Sentence 1 has a higher probability under the bigram MLE model.")
elif prob2 > prob1:
    print("\nThe model prefers Sentence 2.")
    print("Reason: Sentence 2 has a higher probability under the bigram MLE model.")
else:
    print("\nThe model assigns equal probability to both sentences.")
```
